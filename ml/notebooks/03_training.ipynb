# 🧠 CrediShield AI — Model Training, Evaluation & Serialization
## Class-Weighted XGBoost, Stratified Cross-Validation & SHAP Explainability

### End-to-End Pipeline Execution:
1. **Load Data & Leakage Filtering**: enforce Bucket A features only
2. **Data Cleaning**: fix `DAYS_EMPLOYED` anomaly & missing values
3. **Stratified Split**: Train (70%), Validation (15%), Test (15% held out untouched)
4. **ColumnTransformer Preprocessor**: OneHotEncoder (handle_unknown='ignore') + StandardScaler
5. **XGBoost Classifier**: `scale_pos_weight` tuned to class imbalance (~8% default rate)
6. **Evaluation**: ROC-AUC, PR-AUC, Precision, Recall, F1, Confusion Matrix on Test Set
7. **Cost-Based Threshold Sweep**: evaluate 5x False Negative vs False Positive cost
8. **SHAP Explainability**: TreeExplainer native attribution
9. **Artifact Serialization**: save combined preprocessor + model + metadata


```python
import sys
import os
sys.path.append(os.path.abspath('..'))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import shap
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report, confusion_matrix

from src.pipeline import clean_data, build_preprocessor, split_data, get_feature_lists, evaluate_model, compute_shap_values
from src.thresholds import cost_based_sweep, apply_threshold, save_thresholds
from src.serialize import ModelArtifact, save_artifact

print('All modules imported successfully.')
```

```python
# Load data
TRAIN_PATH = '../data/raw/application_train.csv'

if os.path.exists(TRAIN_PATH):
    print(f'Loading raw dataset from {TRAIN_PATH}...')
    df = pd.read_csv(TRAIN_PATH)
else:
    print('Generating synthetic development dataset matching Home Credit schema...')
    np.random.seed(42)
    n = 15000
    df = pd.DataFrame({
        'SK_ID_CURR': np.arange(100000, 100000 + n),
        'TARGET': np.random.choice([0, 1], size=n, p=[0.92, 0.08]),
        'CODE_GENDER': np.random.choice(['M', 'F'], size=n),
        'FLAG_OWN_CAR': np.random.choice(['Y', 'N'], size=n),
        'FLAG_OWN_REALTY': np.random.choice(['Y', 'N'], size=n),
        'CNT_CHILDREN': np.random.poisson(0.5, size=n),
        'AMT_INCOME_TOTAL': np.random.exponential(150000, size=n) + 40000,
        'AMT_CREDIT': np.random.exponential(400000, size=n) + 100000,
        'AMT_ANNUITY': np.random.exponential(25000, size=n) + 5000,
        'NAME_INCOME_TYPE': np.random.choice(['Working', 'Commercial associate', 'Pensioner', 'State servant'], size=n),
        'NAME_EDUCATION_TYPE': np.random.choice(['Higher education', 'Secondary / secondary special'], size=n),
        'NAME_FAMILY_STATUS': np.random.choice(['Married', 'Single / not married'], size=n),
        'NAME_HOUSING_TYPE': np.random.choice(['House / apartment', 'Rented apartment'], size=n),
        'DAYS_BIRTH': -np.random.randint(7500, 25000, size=n),
        'DAYS_EMPLOYED': -np.random.randint(100, 10000, size=n),
        'NAME_CONTRACT_TYPE': np.random.choice(['Cash loans', 'Revolving loans'], size=n),
        'EXT_SOURCE_1': np.random.uniform(0.1, 0.9, size=n),
        'EXT_SOURCE_2': np.random.uniform(0.1, 0.9, size=n),
        'EXT_SOURCE_3': np.random.uniform(0.1, 0.9, size=n),
    })
    # Induce correlations with target
    default_mask = df['TARGET'] == 1
    df.loc[default_mask, 'EXT_SOURCE_2'] *= 0.6
    df.loc[default_mask, 'EXT_SOURCE_3'] *= 0.6
    df.loc[default_mask, 'AMT_CREDIT'] *= 1.4

print(f'Dataset ready: {df.shape[0]:,} rows, {df.shape[1]} columns. Target balance: {df["TARGET"].mean()*100:.2f}% default.')
```

```python
# Preprocess and split
df_clean = clean_data(df)

# Exclude target and identifiers from modeling features
feature_cols = [c for c in df_clean.columns if c not in ['TARGET', 'SK_ID_CURR']]
cat_cols = df_clean[feature_cols].select_dtypes(include=['object', 'category']).columns.tolist()
num_cols = df_clean[feature_cols].select_dtypes(include=[np.number]).columns.tolist()

print(f'Categorical features ({len(cat_cols)}):', cat_cols)
print(f'Numerical features ({len(num_cols)}):', num_cols)

# Stratified 70 / 15 / 15 split
X_train, X_val, X_test, y_train, y_val, y_test = split_data(df_clean, target_col='TARGET', test_size=0.15, val_size=0.15)
print(f'Train samples: {len(X_train):,} | Val: {len(X_val):,} | Test: {len(X_test):,}')
```

```python
# Build and fit preprocessor on Train set only (prevents data snooping / leakage)
preprocessor = build_preprocessor(categorical_features=cat_cols, numerical_features=num_cols)
X_train_trans = preprocessor.fit_transform(X_train[cat_cols + num_cols])
X_val_trans   = preprocessor.transform(X_val[cat_cols + num_cols])
X_test_trans  = preprocessor.transform(X_test[cat_cols + num_cols])

feature_names = preprocessor.get_feature_names_out()
print(f'Transformed feature dimension: {X_train_trans.shape[1]}')
```

```python
# Calculate scale_pos_weight to compensate for ~8% default rate
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
scale_weight = float(neg_count / max(1, pos_count))
print(f'Calculated scale_pos_weight: {scale_weight:.2f}')

# Train XGBoost
model = XGBClassifier(
    n_estimators=200,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_weight,
    random_state=42,
    eval_metric='auc',
    early_stopping_rounds=20,
    n_jobs=-1,
)

model.fit(
    X_train_trans, y_train,
    eval_set=[(X_val_trans, y_val)],
    verbose=20,
)
print('Training complete.')
```

```python
# Untouched Test Set Evaluation
y_test_probs = model.predict_proba(X_test_trans)[:, 1]
roc_auc = roc_auc_score(y_test, y_test_probs)
pr_auc  = average_precision_score(y_test, y_test_probs)

print('=' * 50)
print('UNTOUCHED TEST SET METRICS')
print('=' * 50)
print(f'ROC-AUC Score: {roc_auc:.4f}')
print(f'PR-AUC Score:  {pr_auc:.4f}')
print(classification_report(y_test, (y_test_probs >= 0.5).astype(int)))
```

```python
# Cost-Based Threshold Optimization on Validation Set
y_val_probs = model.predict_proba(X_val_trans)[:, 1]
threshold_results = cost_based_sweep(y_val, y_val_probs, cost_fn_ratio=5.0)

print('=' * 50)
print('COST-BASED THRESHOLD SWEEP RESULTS')
print('=' * 50)
print(f'Optimal Low Threshold (Approved Cutoff):   {threshold_results["low_threshold"]:.4f}')
print(f'Optimal High Threshold (Rejected Cutoff): {threshold_results["high_threshold"]:.4f}')
print(f'Assumed Cost Ratio (FN / FP):             5.0x')

# Export thresholds to backend/policy/thresholds.json
save_thresholds(threshold_results, '../../backend/policy/thresholds.json')
print('Updated backend/policy/thresholds.json with validation-optimized cutoffs.')
```

```python
# SHAP TreeExplainer & Global Summary
explainer = shap.TreeExplainer(model)
sample_idx = np.random.choice(X_test_trans.shape[0], size=min(500, X_test_trans.shape[0]), replace=False)
shap_values = explainer.shap_values(X_test_trans[sample_idx])

plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X_test_trans[sample_idx], feature_names=feature_names, show=False)
os.makedirs('../data/processed', exist_ok=True)
plt.savefig('../data/processed/shap_global_summary.png', dpi=150, bbox_inches='tight')
plt.show()
print('Global SHAP feature attribution plot generated.')
```

```python
# Model Serialization (Full Artifact Bundle)
os.makedirs('../artifacts', exist_ok=True)

artifact = ModelArtifact(
    preprocessor=preprocessor,
    model=model,
    feature_names=list(feature_names),
    categorical_features=cat_cols,
    numerical_features=num_cols,
    model_version='v1.0.0',
    trained_at=pd.Timestamp.now().isoformat(),
    training_metrics={'roc_auc': float(roc_auc), 'pr_auc': float(pr_auc)},
    threshold_policy=threshold_results,
)

save_artifact(artifact, output_dir='../artifacts', name='model_v1')
print('✅ Artifact successfully bundled at ../artifacts/model_v1.joblib and meta sidecar!')
```
