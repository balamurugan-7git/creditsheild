# 🛡️ CrediShield AI — Target Leakage Exclusion Checklist
## Defensible Boundary Between Application-Time & Post-Decision Data

### Purpose & Context
In credit scoring, target leakage is the single most common cause of high offline AUC that collapses in live production. If a model trains on features known only **after** a loan officer approves or services a loan (e.g., late payment tallies, collection activity), it cheats by predicting defaults using consequences of default.

### The Loan Officer Heuristic
> **"Would a human loan officer have known this value at the exact moment before clicking Approve?"**
- **YES** → **Bucket A** (Eligible for modeling)
- **NO**  → **Bucket B** (Excluded as Target Leakage)


```python
import pandas as pd
import numpy as np
import os

print('Initializing Leakage Audit...')
```

```python
# Load columns from raw Home Credit dataset (or schema definition if raw data not yet downloaded)
DATA_PATH = '../data/raw/application_train.csv'
if os.path.exists(DATA_PATH):
    sample_df = pd.read_csv(DATA_PATH, nrows=100)
    columns = sample_df.columns.tolist()
    print(f'Loaded {len(columns)} columns from application_train.csv')
else:
    print('Raw dataset not yet present in data/raw/. Using complete Home Credit specification list.')
    columns = [
        'SK_ID_CURR', 'TARGET', 'NAME_CONTRACT_TYPE', 'CODE_GENDER', 'FLAG_OWN_CAR', 'FLAG_OWN_REALTY',
        'CNT_CHILDREN', 'AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE',
        'NAME_TYPE_SUITE', 'NAME_INCOME_TYPE', 'NAME_EDUCATION_TYPE', 'NAME_FAMILY_STATUS',
        'NAME_HOUSING_TYPE', 'REGION_POPULATION_RELATIVE', 'DAYS_BIRTH', 'DAYS_EMPLOYED',
        'DAYS_REGISTRATION', 'DAYS_ID_PUBLISH', 'OWN_CAR_AGE', 'FLAG_MOBIL', 'FLAG_EMP_PHONE',
        'FLAG_WORK_PHONE', 'FLAG_CONT_MOBILE', 'FLAG_PHONE', 'FLAG_EMAIL', 'OCCUPATION_TYPE',
        'CNT_FAM_MEMBERS', 'REGION_RATING_CLIENT', 'REGION_RATING_CLIENT_W_CITY',
        'WEEKDAY_APPR_PROCESS_START', 'HOUR_APPR_PROCESS_START', 'REG_REGION_NOT_LIVE_REGION',
        'REG_REGION_NOT_WORK_REGION', 'LIVE_REGION_NOT_WORK_REGION', 'REG_CITY_NOT_LIVE_CITY',
        'REG_CITY_NOT_WORK_CITY', 'LIVE_CITY_NOT_WORK_CITY', 'ORGANIZATION_TYPE',
        'EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3',
        'APARTMENTS_AVG', 'BASEMENTAREA_AVG', 'YEARS_BEGINEXPLUATATION_AVG',
        'OBS_30_CNT_SOCIAL_CIRCLE', 'DEF_30_CNT_SOCIAL_CIRCLE',
        'OBS_60_CNT_SOCIAL_CIRCLE', 'DEF_60_CNT_SOCIAL_CIRCLE',
        'DAYS_LAST_PHONE_CHANGE', 'FLAG_DOCUMENT_2', 'FLAG_DOCUMENT_3',
        'AMT_REQ_CREDIT_BUREAU_HOUR', 'AMT_REQ_CREDIT_BUREAU_DAY', 'AMT_REQ_CREDIT_BUREAU_WEEK',
        'AMT_REQ_CREDIT_BUREAU_MON', 'AMT_REQ_CREDIT_BUREAU_QRT', 'AMT_REQ_CREDIT_BUREAU_YEAR'
    ]
```

```python
# Classification checklist logic
checklist = []

# Explicit rules for post-decision / identifier / target
EXCLUDE_EXACT = {
    'SK_ID_CURR': ('Identifier', 'Primary key loan application ID; has no causal predictive validity and risks memorization.'),
    'TARGET': ('Target', 'Ground-truth default label (1 = default, 0 = non-default).'),
}

# Substring or domain-based classification
for col in columns:
    if col in EXCLUDE_EXACT:
        cat, reason = EXCLUDE_EXACT[col]
        checklist.append({'Feature': col, 'Bucket': 'Exclude', 'Category': cat, 'Audit Rationale': reason})
    elif 'DAYS_LAST_DUE' in col or 'PAYMENT' in col or 'DPD' in col:
        checklist.append({'Feature': col, 'Bucket': 'Bucket B (Excluded)', 'Category': 'Post-Approval Performance', 'Audit Rationale': 'Reflects repayment activity post disbursement. Definite target leakage.'})
    elif 'DEF_30' in col or 'DEF_60' in col:
        checklist.append({'Feature': col, 'Bucket': 'Bucket A (Eligible)', 'Category': 'Social Default Observation', 'Audit Rationale': 'Social circle delinquency observation known at application time from external bureau data.'})
    elif col.startswith('AMT_REQ_CREDIT_BUREAU'):
        checklist.append({'Feature': col, 'Bucket': 'Bucket A (Eligible)', 'Category': 'Bureau Inquiries', 'Audit Rationale': 'Enquiry count prior to application; credit seeking behavior known at application.'})
    elif col.startswith('EXT_SOURCE'):
        checklist.append({'Feature': col, 'Bucket': 'Bucket A (Eligible)', 'Category': 'External Bureau Score', 'Audit Rationale': 'Normalized credit bureau scores delivered prior to underwriting decision.'})
    else:
        checklist.append({'Feature': col, 'Bucket': 'Bucket A (Eligible)', 'Category': 'Application Demographics/Financials', 'Audit Rationale': 'Self-reported or verified demographic/loan specification known before decision.'})

checklist_df = pd.DataFrame(checklist)
print(checklist_df['Bucket'].value_counts())
display(checklist_df.head(15))
```

```python
# Export checklist artifact
os.makedirs('../data/processed', exist_ok=True)
checklist_df.to_csv('../data/processed/leakage_checklist.csv', index=False)
try:
    checklist_df.to_excel('../data/processed/leakage_checklist.xlsx', index=False)
    print('✅ Saved audit artifact to ../data/processed/leakage_checklist.xlsx')
except Exception:
    print('✅ Saved audit artifact to ../data/processed/leakage_checklist.csv')
```
