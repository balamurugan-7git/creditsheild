# 📊 CrediShield AI — EDA Notebook
## Home Credit Default Risk Dataset

This notebook performs Exploratory Data Analysis on the Home Credit Default Risk dataset.
**Before running:** ensure `application_train.csv` is in `../data/raw/`


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')

print('Libraries loaded successfully')
```

```python
# Load dataset
TRAIN_PATH = '../data/raw/application_train.csv'
TEST_PATH  = '../data/raw/application_test.csv'

df = pd.read_csv(TRAIN_PATH)
df_test = pd.read_csv(TEST_PATH)

print(f'Train shape: {df.shape}')
print(f'Test shape:  {df_test.shape}')
df.head(3)
```

```python
# ── Basic Info ─────────────────────────────────────────────
print('=== Data Types ===')
print(df.dtypes.value_counts())
print(f'\n=== Memory Usage: {df.memory_usage(deep=True).sum() / 1e6:.1f} MB ===')
```

```python
# ── Target Distribution ────────────────────────────────────
target_counts = df['TARGET'].value_counts()
default_rate = target_counts[1] / len(df) * 100

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].bar(['Non-Default (0)', 'Default (1)'], target_counts.values, color=['#2ecc71', '#e74c3c'])
axes[0].set_title('Class Distribution (Count)', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Count')
for i, v in enumerate(target_counts.values):
    axes[0].text(i, v + 500, f'{v:,}', ha='center', fontweight='bold')

axes[1].pie(target_counts.values, labels=['Non-Default', 'Default'],
            colors=['#2ecc71', '#e74c3c'], autopct='%1.1f%%', startangle=90)
axes[1].set_title(f'Class Distribution (%)\nDefault Rate: {default_rate:.2f}%', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('../data/processed/class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'\n⚠️  Class imbalance detected: {default_rate:.2f}% default rate')
print(f'  scale_pos_weight for XGBoost: {target_counts[0]/target_counts[1]:.2f}')
```

```python
# ── Missing Values ─────────────────────────────────────────
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
missing_df = missing_df[missing_df['Missing Count'] > 0].sort_values('Missing %', ascending=False)

print(f'Columns with missing values: {len(missing_df)}')
print(f'\nTop 20 columns by missing %:')
display(missing_df.head(20))

# Plot top 30 missing
top_missing = missing_df.head(30)
fig, ax = plt.subplots(figsize=(12, 8))
ax.barh(top_missing.index[::-1], top_missing['Missing %'][::-1], color='#e74c3c', alpha=0.8)
ax.set_xlabel('Missing %')
ax.set_title('Top 30 Columns by Missing Data %', fontsize=14, fontweight='bold')
ax.axvline(x=50, color='orange', linestyle='--', label='50% threshold')
ax.legend()
plt.tight_layout()
plt.savefig('../data/processed/missing_values.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Key Numerical Features by Target ──────────────────────
key_num_features = ['AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']
key_num_features = [f for f in key_num_features if f in df.columns]

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for i, col in enumerate(key_num_features):
    ax = axes[i]
    df[df['TARGET']==0][col].dropna().hist(ax=ax, bins=50, alpha=0.6, color='#2ecc71', label='Non-Default', density=True)
    df[df['TARGET']==1][col].dropna().hist(ax=ax, bins=50, alpha=0.6, color='#e74c3c', label='Default', density=True)
    ax.set_title(col, fontweight='bold')
    ax.legend()
    ax.set_ylabel('Density')

plt.suptitle('Key Feature Distributions by Target', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../data/processed/feature_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── EXT_SOURCE Correlation with Target (key predictors) ───
ext_cols = ['EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3']
for col in ext_cols:
    if col in df.columns:
        corr = df[col].corr(df['TARGET'])
        print(f'{col} correlation with TARGET: {corr:.4f}')
```

```python
# ── DAYS_EMPLOYED Anomaly ──────────────────────────────────
print('DAYS_EMPLOYED value_counts for positive values:')
print(df[df['DAYS_EMPLOYED'] > 0]['DAYS_EMPLOYED'].value_counts().head())
print(f'\nRows where DAYS_EMPLOYED == 365243: {(df["DAYS_EMPLOYED"] == 365243).sum()}')
print('→ This sentinel value (365243) represents unemployed applicants. Will replace with NaN.')
```

```python
# ── Categorical Feature Default Rates ─────────────────────
cat_features = ['NAME_CONTRACT_TYPE', 'CODE_GENDER', 'NAME_INCOME_TYPE',
                'NAME_EDUCATION_TYPE', 'NAME_FAMILY_STATUS', 'NAME_HOUSING_TYPE']
cat_features = [f for f in cat_features if f in df.columns]

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

for i, col in enumerate(cat_features):
    ax = axes[i]
    dr = df.groupby(col)['TARGET'].mean().sort_values(ascending=False) * 100
    colors = ['#e74c3c' if v > df['TARGET'].mean()*100 else '#2ecc71' for v in dr.values]
    bars = ax.bar(range(len(dr)), dr.values, color=colors)
    ax.set_xticks(range(len(dr)))
    ax.set_xticklabels(dr.index, rotation=45, ha='right', fontsize=9)
    ax.set_title(f'Default Rate by {col}', fontweight='bold')
    ax.set_ylabel('Default Rate (%)')
    ax.axhline(y=df['TARGET'].mean()*100, color='orange', linestyle='--', label=f'Overall ({df["TARGET"].mean()*100:.1f}%)')
    ax.legend(fontsize=8)

plt.suptitle('Default Rates by Categorical Features', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('../data/processed/categorical_default_rates.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# ── Correlation Heatmap (numerical features) ───────────────
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
num_cols = [c for c in num_cols if c != 'SK_ID_CURR' and df[c].notna().sum() > len(df)*0.5]

corr_with_target = df[num_cols].corr()['TARGET'].drop('TARGET').abs().sort_values(ascending=False)
print('Top 15 features by absolute correlation with TARGET:')
display(corr_with_target.head(15).to_frame('|correlation|'))

# Check for potential leakage (correlation > 0.9)
high_corr = corr_with_target[corr_with_target > 0.9]
if len(high_corr) > 0:
    print(f'\n⚠️  HIGH CORRELATION (>0.9) FEATURES — potential leakage!:')
    print(high_corr)
else:
    print('\n✅ No features with |correlation| > 0.9 — no obvious leakage.')
```
